# Transforming a direction between spacecraft frames

A star tracker frame $S$ is mounted to the spacecraft body frame $B$. The unit quaternion $q_{BS}$ describes the orientation of $S$ relative to $B$ and maps a vector expressed in $S$ into coordinates in $B$.

Suppose the tracker is rotated by $\theta=30^\circ$ about the positive $y$-axis. A direction measured in the tracker frame is

$$
\mathbf{d}_S = \begin{bmatrix}0 \\ 0 \\ 1\end{bmatrix}.
$$

This notebook constructs the rotation quaternion, applies it to the direction, and then applies the inverse rotation to check that the original direction is recovered. Quaternions are written scalar-first: $q=[q_0,q_1,q_2,q_3]$, with the last three components corresponding to the $i,j,k$ axes.

```python
from aocs.common.quaternion import Quaternion
```

```python
d_s = Quaternion.from_vector([0, 0, 1])
d_s
```

For a rotation by $\theta$ about the unit axis $\hat{\mathbf{u}}=(u_x,u_y,u_z)$, the scalar-first rotation quaternion is

$$
q_{BS} = \begin{bmatrix}\cos(\theta/2) \\ u_x\sin(\theta/2) \\ u_y\sin(\theta/2) \\ u_z\sin(\theta/2)\end{bmatrix}.
$$

Here $\hat{\mathbf{u}}=(0,1,0)$ and $\theta=30^\circ=\pi/6$ radians, so

$$
q_{BS} = \begin{bmatrix}\cos(15^\circ) \\ 0 \\ \sin(15^\circ) \\ 0\end{bmatrix}.
$$

The implementation expects angles in radians and normalises the supplied axis.

```python
from math import radians

q_BS = Quaternion.from_axis_angle(theta=radians(30), axis=[0, 1, 0])
q_BS
```

```python
d_b = q_BS * d_s * q_BS.conjugate()
d_b
```

Embed the vector as a pure quaternion, $\tilde{d}_S=[0,\mathbf{d}_S]$. Rotate it into frame $B$ with

$$
\tilde{d}_B = q_{BS}\,\tilde{d}_S\,q_{BS}^{*},
$$

where $q_{BS}^{*}$ is the conjugate. Because $q_{BS}$ is a unit quaternion, its conjugate is also its inverse. A positive $30^\circ$ rotation about $+y$ takes the $+z$ direction toward $+x$, giving $\mathbf{d}_B=(\sin 30^\circ,0,\cos 30^\circ)=(0.5,0,\sqrt{3}/2)$.

```python
d_s_check = q_BS.conjugate() * d_b * q_BS
d_s_check
```

```python
assert d_s.is_close(d_s_check)
```
